```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt

# 2. MNIST dataset load karna (Digits 0 to 9)
(train_images, train_labels), (test_images, test_labels) = datasets.mnist.load_data()

# Images ko 0 se 1 ke beech normalize karna
train_images = train_images / 255.0
test_images = test_images / 255.0

# Dataset ko CNN ke liye sahi shape me reshape karna
train_images = train_images.reshape((60000, 28, 28, 1))
test_images = test_images.reshape((10000, 28, 28, 1))

print("Dataset successfully loaded and preprocessed!")

# 3. Convolutional Neural Network (CNN) Model Banana
model = models.Sequential()
model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))

# Dense layers add karna classification ke liye
model.add(layers.Flatten())
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(10, activation='softmax')) # 0 se 9 tak 10 digits hote hain

# 4. Model Compile Karna
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 5. Model Train Karna (Epochs = 5 rakh rahe hain taaki jaldi run ho jaye)
history = model.fit(train_images, train_labels, epochs=5, 
                    validation_data=(test_images, test_labels))

# 6. Test data par Accuracy Check karna
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
print(f"\nModel Test Accuracy: {test_acc * 100:.2f}%")
```
